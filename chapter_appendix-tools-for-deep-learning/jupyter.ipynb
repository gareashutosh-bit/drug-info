```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix
```

```python
#Actual Array VS Predicted Array
Actual=np.array(['CAT','CAT','CAT','NOTCAT','CAT','NOTCAT','CAT','CAT','CAT','CAT'])
Predicted=np.array(['CAT','NOTCAT','CAT','CAT','CAT','NOTCAT','CAT','CAT','NOTCAT','CAT'])
```

```python
#Compute Confusion Matrix
conf_matrix=confusion_matrix(Actual,Predicted)
```

```python
sns.heatmap(conf_matrix,annot=True,xticklabels=['CAT','NOTCAT'],yticklabels=['CAT','NOTCAT'])
```
